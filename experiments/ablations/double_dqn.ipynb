```python

# ============================================================
# 1. Economic environment + DQN components
# ============================================================

R = 1.00985
BETA = 0.9703
SIGMA = 2.0

Y_HIGH = 1.0
Y_LOW = 0.472

P_INCOME = np.array(
    [
        [0.9393, 0.0607],  # High -> High, Low
        [0.3920, 0.6080],  # Low  -> High, Low
    ],
    dtype=np.float64
)

SAVING_SHARES = np.arange(
    0.0, 1.0, 0.05, dtype=np.float64
)


class ConsumptionSavingEnv(gym.Env):
    """
    Infinite-horizon consumption-saving environment.

    State:
        [asset_t, income_t]

    Action:
        discrete saving-share index
        0 -> 0.00, 1 -> 0.05, ..., 19 -> 0.95

    Reward:
        CRRA utility of current consumption

    Income state:
        0 = High
        1 = Low
    """

    metadata = {"render_modes": []}

    def __init__(
        self,
        R=R,
        beta=BETA,
        sigma=SIGMA,
        y_high=Y_HIGH,
        y_low=Y_LOW,
        transition_matrix=P_INCOME,
        initial_asset=1.0,
        initial_income_state=0,
    ):
        super().__init__()

        self.R = float(R)
        self.beta = float(beta)
        self.sigma = float(sigma)

        self.income_values = np.array(
            [y_high, y_low],
            dtype=np.float64
        )

        self.P = np.asarray(
            transition_matrix,
            dtype=np.float64
        )

        if self.P.shape != (2, 2):
            raise ValueError("transition_matrix must be (2,2).")

        if not np.allclose(self.P.sum(axis=1), 1.0):
            raise ValueError("Each row of transition_matrix must sum to 1.")

        self.initial_asset = float(initial_asset)
        self.initial_income_state = int(initial_income_state)

        self.saving_shares = SAVING_SHARES.copy()

        self.action_space = spaces.Discrete(
            len(self.saving_shares)
        )

        self.observation_space = spaces.Box(
            low=np.array([0.0, min(y_low, y_high)], dtype=np.float32),
            high=np.array([np.inf, max(y_low, y_high)], dtype=np.float32),
            dtype=np.float32
        )

        self.asset = None
        self.income_state = None

    def _get_obs(self):
        return np.array(
            [
                self.asset,
                self.income_values[self.income_state]
            ],
            dtype=np.float32
        )

    def _utility(self, consumption):
        if consumption <= 0:
            raise ValueError("Consumption must be positive.")

        if np.isclose(self.sigma, 1.0):
            return float(np.log(consumption))

        return float(
            consumption ** (1.0 - self.sigma)
            / (1.0 - self.sigma)
        )

    def reset(self, *, seed=None, options=None):
        super().reset(seed=seed)

        options = options or {}

        self.asset = float(
            options.get("initial_asset", self.initial_asset)
        )

        self.income_state = int(
            options.get(
                "initial_income_state",
                self.initial_income_state
            )
        )

        return self._get_obs(), {}

    def step(self, action):
        current_asset = self.asset
        current_income_state = self.income_state
        current_income = self.income_values[current_income_state]

        cash_on_hand = (
            self.R * current_asset
            + current_income
        )

        saving_share = self.saving_shares[action]

        next_asset = (
            saving_share * cash_on_hand
        )

        consumption = (
            cash_on_hand - next_asset
        )

        reward = self._utility(consumption)

        next_income_state = int(
            self.np_random.choice(
                2,
                p=self.P[current_income_state]
            )
        )

        self.asset = next_asset
        self.income_state = next_income_state

        # Infinite-horizon model: no economic terminal state
        terminated = False
        truncated = False

        return (
            self._get_obs(),
            reward,
            terminated,
            truncated,
            {}
        )


def normalize_state(state):
    state = np.asarray(
        state,
        dtype=np.float32
    ).copy()

    state[..., 0] /= 4.5

    return state


class ReplayBuffer:

    def __init__(self, capacity=100_000):
        self.buffer = deque(maxlen=capacity)

    def put(self, state, action, reward, next_state):
        self.buffer.append(
            (
                normalize_state(state),
                int(action),
                float(reward),
                normalize_state(next_state)
            )
        )

    def sample(self, batch_size):
        batch = random.sample(
            self.buffer,
            batch_size
        )

        states, actions, rewards, next_states = zip(*batch)

        return (
            torch.tensor(
                np.asarray(states),
                dtype=torch.float32
            ),
            torch.tensor(
                actions,
                dtype=torch.long
            ),
            torch.tensor(
                rewards,
                dtype=torch.float32
            ),
            torch.tensor(
                np.asarray(next_states),
                dtype=torch.float32
            )
        )

    def __len__(self):
        return len(self.buffer)


class QNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.fc1 = nn.Linear(2, 64)
        self.fc2 = nn.Linear(64, 64)
        self.fc3 = nn.Linear(64, 20)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))

        return self.fc3(x)


print("Core definitions loaded.")
```

```python

# ============================================================
# 2. DP-Matched benchmark
# ============================================================

def solve_dp_matched(
    R=R,
    beta=BETA,
    sigma=SIGMA,
    y_high=Y_HIGH,
    y_low=Y_LOW,
    transition_matrix=P_INCOME,
    a_min=0.0,
    a_max=4.5,
    n_assets=500,
    saving_shares=SAVING_SHARES,
    tolerance=1e-10,
    max_iterations=5000
):
    P = np.asarray(
        transition_matrix,
        dtype=np.float64
    )

    income_values = np.array(
        [y_high, y_low],
        dtype=np.float64
    )

    asset_grid = np.linspace(
        a_min,
        a_max,
        n_assets
    )

    V = np.zeros(
        (n_assets, 2),
        dtype=np.float64
    )

    policy_share = np.zeros_like(V)

    for iteration in range(max_iterations):

        V_new = np.empty_like(V)

        for income_state in range(2):

            income = income_values[income_state]

            cash_on_hand = (
                R * asset_grid + income
            )

            Q = np.full(
                (n_assets, len(saving_shares)),
                -np.inf,
                dtype=np.float64
            )

            for j, z in enumerate(saving_shares):

                next_asset = (
                    z * cash_on_hand
                )

                consumption = (
                    (1.0 - z) * cash_on_hand
                )

                feasible = (
                    (consumption > 1e-12)
                    &
                    (next_asset <= a_max)
                )

                V_next_high = np.interp(
                    next_asset,
                    asset_grid,
                    V[:, 0]
                )

                V_next_low = np.interp(
                    next_asset,
                    asset_grid,
                    V[:, 1]
                )

                expected_future = (
                    P[income_state, 0] * V_next_high
                    +
                    P[income_state, 1] * V_next_low
                )

                if np.isclose(sigma, 1.0):
                    utility = np.log(consumption)
                else:
                    utility = (
                        consumption ** (1.0 - sigma)
                        /
                        (1.0 - sigma)
                    )

                q = (
                    utility
                    +
                    beta * expected_future
                )

                q[~feasible] = -np.inf
                Q[:, j] = q

            best_idx = np.argmax(
                Q,
                axis=1
            )

            V_new[:, income_state] = (
                Q[np.arange(n_assets), best_idx]
            )

            policy_share[:, income_state] = (
                saving_shares[best_idx]
            )

        diff = np.max(
            np.abs(V_new - V)
        )

        V = V_new

        if diff < tolerance:
            break

    return asset_grid, V, policy_share


asset_grid_m, V_dp_matched, policy_share_matched = (
    solve_dp_matched()
)

dp_matched_initial_value = np.interp(
    1.0,
    asset_grid_m,
    V_dp_matched[:, 0]
)

low_asset_mask = (
    asset_grid_m <= 0.5
)

middle_asset_mask = (
    (asset_grid_m > 0.5)
    &
    (asset_grid_m <= 2.0)
)

high_asset_mask = (
    asset_grid_m > 2.0
)

print(
    "DP-Matched Bellman value at (asset=1, High):",
    dp_matched_initial_value
)

print("DP-Matched benchmark ready.")
```

```python

# ============================================================
# 3. Evaluation helpers
# ============================================================

def utility_vectorized(consumption, sigma=SIGMA):
    if np.isclose(sigma, 1.0):
        return np.log(consumption)

    return (
        consumption ** (1.0 - sigma)
        /
        (1.0 - sigma)
    )


def evaluate_dqn_vectorized(
    q_network,
    n_episodes=2000,
    horizon=500,
    seed=10000,
    R=R,
    beta=BETA,
    sigma=SIGMA,
    y_high=Y_HIGH,
    y_low=Y_LOW
):
    rng = np.random.default_rng(seed)

    assets = np.ones(n_episodes)

    income_states = np.zeros(
        n_episodes,
        dtype=int
    )

    total_utility = np.zeros(n_episodes)

    discount = 1.0

    q_network.eval()

    for _ in range(horizon):

        incomes = np.where(
            income_states == 0,
            y_high,
            y_low
        )

        states = np.column_stack(
            [assets, incomes]
        ).astype(np.float32)

        states[:, 0] /= 4.5

        with torch.no_grad():

            actions = (
                q_network(
                    torch.tensor(
                        states,
                        dtype=torch.float32
                    )
                )
                .argmax(dim=1)
                .cpu()
                .numpy()
            )

        z = SAVING_SHARES[actions]

        cash_on_hand = (
            R * assets + incomes
        )

        consumption = (
            (1.0 - z)
            * cash_on_hand
        )

        reward = utility_vectorized(
            consumption,
            sigma
        )

        total_utility += (
            discount * reward
        )

        assets = (
            z * cash_on_hand
        )

        prob_low = np.where(
            income_states == 0,
            P_INCOME[0, 1],
            P_INCOME[1, 1]
        )

        income_states = (
            rng.random(n_episodes)
            < prob_low
        ).astype(int)

        discount *= beta

    mean = total_utility.mean()
    std = total_utility.std(ddof=1)
    se = std / np.sqrt(n_episodes)
    ci = 1.96 * se

    return {
        "mean_utility": mean,
        "std_utility": std,
        "ci_lower": mean - ci,
        "ci_upper": mean + ci,
        "max_final_asset": assets.max()
    }


def get_dqn_policy_on_grid(
    q_network,
    asset_grid
):
    q_network.eval()

    policies = np.zeros(
        (len(asset_grid), 2)
    )

    income_values = [
        Y_HIGH,
        Y_LOW
    ]

    for income_state, income in enumerate(
        income_values
    ):

        states = np.column_stack(
            [
                asset_grid,
                np.full(
                    len(asset_grid),
                    income
                )
            ]
        ).astype(np.float32)

        states[:, 0] /= 4.5

        with torch.no_grad():

            actions = (
                q_network(
                    torch.tensor(
                        states,
                        dtype=torch.float32
                    )
                )
                .argmax(dim=1)
                .cpu()
                .numpy()
            )

        policies[:, income_state] = (
            SAVING_SHARES[actions]
        )

    return policies


print("Evaluation helpers loaded.")
```


# Next Experiment — Double DQN

이번 실험은 checkpoint 결과에서 확인된 **DQN training instability**를 직접 겨냥합니다.

기존 `v1-long`과 동일하게 유지:
- 100,000 steps
- learning rate = 0.001
- MSE loss
- target update = 500
- epsilon decay = 25,000
- initial asset ~ Uniform(0, 4.5)
- same network / replay buffer / normalization

**유일한 알고리즘 변경: TD target을 Double DQN 방식으로 계산합니다.**

기존 DQN:
\[
y=r+\gamma\max_a Q_{\text{target}}(s',a)
\]

Double DQN:
\[
a^*=\arg\max_a Q_{\text{online}}(s',a)
\]

\[
y=r+\gamma Q_{\text{target}}(s',a^*)
\]

목적:
- max 연산으로 인한 Q-value overestimation을 줄이는지
- 학습 중 성능 drift가 감소하는지
- DP-Matched와의 welfare / policy gap이 줄어드는지 확인


```python

def train_double_dqn_checkpointed(
    total_steps=100_000,
    seed=42,
    learning_rate=1e-3,
    gamma=BETA,
    batch_size=64,
    buffer_size=100_000,
    warmup_steps=1_000,
    target_update_interval=500,
    epsilon_start=1.0,
    epsilon_end=0.05,
    epsilon_decay_steps=25_000,
    reset_interval=200,
    eval_interval=5_000,
    validation_episodes=512,
    validation_horizon=300,
    validation_seed=54321
):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    rng = np.random.default_rng(seed)

    env = ConsumptionSavingEnv()

    q = QNetwork()
    target_q = QNetwork()
    target_q.load_state_dict(q.state_dict())
    target_q.eval()

    optimizer = optim.Adam(
        q.parameters(),
        lr=learning_rate
    )

    replay_buffer = ReplayBuffer(
        capacity=buffer_size
    )

    def sample_initial_state():
        asset = rng.uniform(0.0, 4.5)
        income_state = int(rng.integers(0, 2))
        return asset, income_state

    a0, y0 = sample_initial_state()

    state, _ = env.reset(
        seed=seed,
        options={
            "initial_asset": a0,
            "initial_income_state": y0
        }
    )

    losses = []
    history = []
    episode_step = 0

    best_validation_utility = -np.inf
    best_step = None
    best_state_dict = None

    for step in range(1, total_steps + 1):

        fraction = min(
            (step - 1) / epsilon_decay_steps,
            1.0
        )

        epsilon = (
            epsilon_start
            + fraction * (epsilon_end - epsilon_start)
        )

        if rng.random() < epsilon:
            action = int(rng.integers(0, 20))
        else:
            s = torch.tensor(
                normalize_state(state),
                dtype=torch.float32
            ).unsqueeze(0)

            with torch.no_grad():
                action = int(
                    q(s)
                    .argmax(dim=1)
                    .item()
                )

        next_state, reward, _, _, _ = env.step(action)

        replay_buffer.put(
            state,
            action,
            reward,
            next_state
        )

        state = next_state
        episode_step += 1

        if episode_step >= reset_interval:
            a0, y0 = sample_initial_state()

            state, _ = env.reset(
                options={
                    "initial_asset": a0,
                    "initial_income_state": y0
                }
            )

            episode_step = 0

        if len(replay_buffer) >= warmup_steps:

            (
                states,
                actions,
                rewards,
                next_states
            ) = replay_buffer.sample(batch_size)

            q_selected = (
                q(states)
                .gather(
                    1,
                    actions.unsqueeze(1)
                )
                .squeeze(1)
            )

            with torch.no_grad():

                # ==========================================
                # Double DQN:
                # online network SELECTS the next action
                # ==========================================
                next_actions = (
                    q(next_states)
                    .argmax(dim=1, keepdim=True)
                )

                # ==========================================
                # target network EVALUATES that action
                # ==========================================
                next_q_double = (
                    target_q(next_states)
                    .gather(1, next_actions)
                    .squeeze(1)
                )

                target = (
                    rewards
                    + gamma * next_q_double
                )

            loss = F.mse_loss(
                q_selected,
                target
            )

            optimizer.zero_grad()
            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                q.parameters(),
                max_norm=10.0
            )

            optimizer.step()

            losses.append(loss.item())

        if step % target_update_interval == 0:
            target_q.load_state_dict(
                q.state_dict()
            )

        if step % eval_interval == 0:

            validation = evaluate_dqn_vectorized(
                q,
                n_episodes=validation_episodes,
                horizon=validation_horizon,
                seed=validation_seed
            )

            policy = get_dqn_policy_on_grid(
                q,
                asset_grid_m
            )

            err = np.abs(
                policy - policy_share_matched
            )

            row = {
                "step": step,
                "validation_utility":
                    validation["mean_utility"],
                "policy_mae":
                    err.mean(),
                "exact":
                    np.mean(
                        np.isclose(
                            policy,
                            policy_share_matched
                        )
                    ),
                "within5":
                    np.mean(
                        err <= 0.0500001
                    ),
                "within10":
                    np.mean(
                        err <= 0.1000001
                    ),
                "epsilon":
                    epsilon
            }

            history.append(row)

            if (
                validation["mean_utility"]
                > best_validation_utility
            ):
                best_validation_utility = (
                    validation["mean_utility"]
                )
                best_step = step

                best_state_dict = {
                    k: v.detach().cpu().clone()
                    for k, v in q.state_dict().items()
                }

            print(
                f"step={step:6d} | "
                f"val utility={validation['mean_utility']:.4f} | "
                f"MAE={err.mean():.4f} | "
                f"best={best_step}"
            )

    best_q = QNetwork()
    best_q.load_state_dict(best_state_dict)
    best_q.eval()

    return (
        q,
        best_q,
        target_q,
        losses,
        pd.DataFrame(history),
        best_step
    )
```
